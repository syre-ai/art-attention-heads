# RD1: ART Attention Pattern Taxonomy — Exploration

Interactive analysis of attention features and ART clustering results.

```python
import sys
sys.path.insert(0, '..')

import numpy as np
import matplotlib.pyplot as plt
from src.extraction.cache_manager import load_features
from src.extraction.feature_computer import FEATURE_NAMES
from src.clustering.art_runner import run_art
from src.validation.known_heads import classify_heads_heuristic

%matplotlib inline
plt.rcParams['figure.dpi'] = 120
```

## Load Features

```python
data = load_features('base')
X = data['aggregated']
per_input = data['per_input']
print(f'Aggregated: {X.shape}')
print(f'Per-input: {per_input.shape}')
print(f'Feature ranges: {X.min():.3f} to {X.max():.3f}')
```

## Feature Distributions

```python
fig, axes = plt.subplots(4, 4, figsize=(16, 12))
for i, (ax, name) in enumerate(zip(axes.flat, FEATURE_NAMES)):
    ax.hist(X[:, i], bins=30, edgecolor='black', alpha=0.7)
    ax.set_title(name, fontsize=10)
    ax.set_xlim(0, 1)
for ax in axes.flat[len(FEATURE_NAMES):]:
    ax.set_visible(False)
plt.suptitle('Feature Distributions (256 heads)', fontsize=14)
plt.tight_layout()
plt.show()
```

## Quick ART Clustering

```python
# Try a few rho values
for rho in [0.3, 0.5, 0.7]:
    result = run_art(X, 'fuzzy_art', rho)
    print(f'rho={rho}: {result.n_clusters} clusters')
```

## Heuristic Head Classification

```python
heuristic_labels, groups = classify_heads_heuristic(X)
for head_type, indices in groups.items():
    if indices:
        print(f'{head_type}: {len(indices)} heads — {indices[:10]}...')
```
